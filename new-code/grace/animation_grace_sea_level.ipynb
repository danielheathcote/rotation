# Animation: sea-level change and the GRACE load

Left: the direct load each month (mm of water equivalent, relative to the 2004–2009 mean), the input to the solver. Middle: the dynamic
sea-level change over the ocean. Right: the dynamic pole path. Frames are the observed months, so the gaps in GRACE appear as jumps. The load's
colour scale is linear within ±50 mm and logarithmic beyond, so that the ocean and land water (tens of mm) and the ice sheets (metres) both show.
The sea-level colour limit ignores the most extreme 1% of values.

```python
import sys
import os
import numpy as np
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
from matplotlib.animation import PillowWriter
from IPython.display import Image

# solver_dynamic and examples_dynamic_helper live one directory up, in new-code/
sys.path.append("..")

import solver_dynamic as sd
import examples_dynamic_helper as eh
import grace_data_helper as gh
import grace_plot_helper as ph
```

```python
#=======================================================================
# Parameters
#=======================================================================

# The run to plot (see grace_rotation_run.py): forcing, degree and quality factor
forcing = "land-ocean"
lmax = 64
quality_factor = "100"

# The load's colour scale is linear within +-load_linear_mm and logarithmic beyond
load_linear_mm = 50.0

frames_per_second = 8
dpi = 80
gif_path = f"animations/grace_sea_level_{forcing}.gif"
```

```python
#=======================================================================
# Earth model, the run and the direct load
#=======================================================================

state = sd.make_state(lmax)
model = state.model
constants = sd.rotation_constants(model)

run = gh.load_run(forcing, lmax, quality_factor)
time_days, dates, load_coeffs = gh.direct_load_coefficients(forcing, lmax)
assert np.allclose(run["month_days"], time_days), "Plot a run that covers every month"

dynamic_mas = sd.omega_to_mas(run["dynamic_omega"], constants)
static_mas = sd.omega_to_mas(run["static_omega"], constants)

# The dynamic pole at every time step, and the step of each month (the frames)
path_mas = run["pole_mas"]
frames = ph.frame_index(run["t_days"], run["month_days"])
print(f"{len(dates)} months, {dates[0]} to {dates[-1]}")
```

```python
#=======================================================================
# Load and sea level at every month (mm)
#=======================================================================

load_mm = ph.fields_from_coefficients(model, load_coeffs)
sea_level_mm = ph.fields_from_coefficients(model, run["dynamic_sea_level_coeffs"], scale=1e3, mask=ph.ocean_mask(state))
print(f"largest load {np.nanmax(np.abs(load_mm)):.0f} mm w.e.; sea-level colour limit {ph.robust_limit(sea_level_mm):.1f} mm")
```

```python
#=======================================================================
# Animation
#=======================================================================

animation = ph.map_animation(
    state,
    [load_mm, sea_level_mm],
    ["Direct load", "Sea-level change"],
    ["Direct load (mm water equivalent)", r"$\xi$ (mm)"],
    ["BrBG", "RdBu_r"],
    [ph.load_norm(load_mm, linear_mm=load_linear_mm), ph.symmetric_norm(sea_level_mm)],
    dates,
    path_mas,
    frames,
    frames_per_second=frames_per_second,
)

os.makedirs(os.path.dirname(gif_path), exist_ok=True)
animation.save(gif_path, writer=PillowWriter(fps=frames_per_second), dpi=dpi)
print(f"saved {gif_path}")
Image(url=gif_path)
```
